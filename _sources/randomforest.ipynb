## Input Data: Excel Reader

Tahap pertama adalah memasukkan data mentah ke dalam KNIME. Node ini membaca file tugas-pohon.xlsx yang berisi histori data cuaca. Kita dapat melihat variabel seperti Outlook (cuaca), Temperature, Humidity, dan Windy. Target yang ingin kita tebak adalah kolom Play Tennis (Yes/No).

## Random Forest Learner

Di sinilah proses "belajar" yang sebenarnya terjadi dengan membangun sekumpulan pohon keputusan.

![alt text](Gambar/trainrandom.png)  

Pada konfigurasi Random Forest Learner, model mencoba mencari hubungan antara cuaca dan keputusan bermain. Algoritma ini tidak hanya mengandalkan satu pohon, melainkan banyak pohon acak untuk memastikan prediksi yang dihasilkan lebih stabil dan tidak hanya "menghafal" data (overfitting).

**Detail Statistik & Validasi:**

![alt text](Gambar/attributestatics.png)  

Tabel Attribute Statistics menunjukkan fitur cuaca mana yang paling sering digunakan dalam pengambilan keputusan. Ini membantu kita memahami faktor terpenting


![alt text](Gambar/outofbag.png) 

Hasil Out-of-Bag (OOB) Error merupakan fitur spesial dari Random Forest. Gambar ini menampilkan "rapor" sementara model, di mana ia mengetes dirinya sendiri menggunakan sebagian data yang tidak terpilih saat proses belajar.
