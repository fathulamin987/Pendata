#### Perhitungan Distance (Euclidean, Manhattan, Minkowski)

#### 1. Representasi Data dalam Bentuk Vektor

Diberikan 3 data dengan dua atribut numerik: Age dan Family Size.

$$
x_1 = (20,4)
$$

$$
x_2 = (24,3)
$$

$$
x_3 = (22,3)
$$

Setiap data direpresentasikan sebagai:

$$
x = (x_1, x_2)
$$

Penjelasan:  
Komponen pertama menyatakan umur, dan komponen kedua menyatakan jumlah anggota keluarga.

---

#### 2. Euclidean Distance

**Rumus**

$$
d(x,y) = \sqrt{(x_1 - y_1)^2 + (x_2 - y_2)^2}
$$

Penjelasan:  
Rumus ini menghitung jarak garis lurus antara dua titik dengan mengkuadratkan selisih tiap atribut, menjumlahkannya, lalu diakar.

**Contoh Perhitungan $x_1$ dan $x_2$**

$$
d(x_1,x_2) = \sqrt{(20 - 24)^2 + (4 - 3)^2}
$$

$$
= \sqrt{(-4)^2 + (1)^2}
$$

$$
= \sqrt{16 + 1}
$$

$$
= \sqrt{17}
$$

$$
\approx 4.123
$$

Artinya jarak antara data 1 dan data 2 adalah sekitar 4.123.

**Matriks Jarak Euclidean**

$$
D =
\begin{bmatrix}
0 & 4.123 & 2.236 \\
4.123 & 0 & 2.000 \\
2.236 & 2.000 & 0
\end{bmatrix}
$$

---

#### 3. Manhattan Distance

**Rumus**

$$
d(x,y) = |x_1 - y_1| + |x_2 - y_2|
$$

Penjelasan:  
Menghitung jumlah selisih absolut tiap atribut tanpa kuadrat dan akar.

**Contoh Perhitungan $x_1$ dan $x_2$**

$$
d(x_1,x_2) = |20 - 24| + |4 - 3|
$$

$$
= 4 + 1
$$

$$
= 5.0
$$

Artinya jarak Manhattan antara data 1 dan data 2 adalah 5.0.

**Matriks Jarak Manhattan**

$$
D =
\begin{bmatrix}
0 & 5.0 & 3.0 \\
5.0 & 0 & 2.0 \\
3.0 & 2.0 & 0
\end{bmatrix}
$$

---

#### 4. Minkowski Distance ($p = 3$)

**Rumus**

$$
d(x,y) = \left( \sum_{i=1}^{n} |x_i - y_i|^p \right)^{\frac{1}{p}}
$$

Dengan:

$$
p = 3
$$

Penjelasan:  
Minkowski adalah bentuk umum dari Manhattan dan Euclidean.  
Jika $p=1$ → Manhattan  
Jika $p=2$ → Euclidean  
Jika $p=3$ → pangkat tiga lalu diakar tiga.

**Contoh Perhitungan $x_1$ dan $x_2$**

$$
d(x_1,x_2) = (|20 - 24|^3 + |4 - 3|^3)^{1/3}
$$

$$
= (|-4|^3 + 1^3)^{1/3}
$$

$$
= (64 + 1)^{1/3}
$$

$$
= \sqrt[3]{65}
$$

$$
\approx 4.021
$$

Artinya jarak Minkowski antara data 1 dan 2 adalah 4.021.

**Matriks Jarak Minkowski**

$$
D =
\begin{bmatrix}
0 & 4.021 & 2.080 \\
4.021 & 0 & 2.000 \\
2.080 & 2.000 & 0
\end{bmatrix}
$$

---

#### Analisis Singkat

1. Matriks bersifat simetris:

$$
d(x_i,x_j) = d(x_j,x_i)
$$

2. Elemen diagonal bernilai nol:

$$
d(x_i,x_i) = 0
$$

3. Data yang paling mirip adalah $x_2$ dan $x_3$ karena memiliki nilai jarak terkecil pada semua metode.

---

**Kesimpulan**

- Euclidean → jarak garis lurus  
- Manhattan → jumlah selisih absolut  
- Minkowski → bentuk umum dengan parameter $p$  
- Data 2 dan Data 3 adalah pasangan paling mirip
