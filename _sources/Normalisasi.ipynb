# Materi Normalisasi
Normalisasi adalah teknik dalam **data preprocessing** yang digunakan untuk mengubah nilai atribut numerik ke dalam skala tertentu tanpa mengubah informasi dasar dari data tersebut.

Tujuan normalisasi:

- Menyamakan skala antar atribut
- Menghindari dominasi atribut bernilai besar
- Meningkatkan akurasi algoritma berbasis jarak



## Normalisasi Min-Max
Normalisasi Min-Max adalah metode normalisasi yang mengubah nilai suatu atribut ke dalam rentang tertentu, biasanya **0 sampai 1**.

Metode ini bekerja dengan cara memetakan nilai dari rentang awal ke rentang baru sehingga seluruh data memiliki skala yang sama.

---

#### Rumus

$$
X' = \frac{X - X_{min}}{X_{max} - X_{min}}
$$

**Keterangan**

- $X$ = nilai asli
- $X'$ = nilai hasil normalisasi
- $X_{min}$ = nilai minimum pada atribut
- $X_{max}$ = nilai maksimum pada atribut

---

#### Dataset

| No | IPK | PO |
|---|---|---|
|1|2|200000|
|2|3|300000|
|3|4|200000|
|4|2|200000|
|5|3|300000|
|6|4|400000|
|7|2|300000|

---

#### Menentukan Nilai Minimum dan Maksimum

**Atribut IPK**

Data: 2, 3, 4, 2, 3, 4, 2  

- Nilai minimum = 2  
- Nilai maksimum = 4  

**Atribut PO**

Data: 200000, 300000, 200000, 200000, 300000, 400000, 300000  

- Nilai minimum = 200000  
- Nilai maksimum = 400000  

---

#### Perhitungan Normalisasi IPK

**Data ke-1 (IPK = 2)**

$$
X' = \frac{2 - 2}{4 - 2}
$$

$$
X' = \frac{0}{2} = 0
$$

---

**Data ke-2 (IPK = 3)**

$$
X' = \frac{3 - 2}{4 - 2}
$$

$$
X' = \frac{1}{2} = 0.5
$$

---

**Data ke-3 (IPK = 4)**

$$
X' = \frac{4 - 2}{4 - 2}
$$

$$
X' = \frac{2}{2} = 1
$$

---

#### Hasil Normalisasi IPK

| No | IPK | IPK Normalisasi |
|---|---|---|
|1|2|0|
|2|3|0.5|
|3|4|1|
|4|2|0|
|5|3|0.5|
|6|4|1|
|7|2|0|

---

#### Perhitungan Normalisasi PO

**Data ke-1 (PO = 200000)**

$$
X' = \frac{200000 - 200000}{400000 - 200000}
$$

$$
X' = \frac{0}{200000} = 0
$$

---

**Data ke-2 (PO = 300000)**

$$
X' = \frac{300000 - 200000}{400000 - 200000}
$$

$$
X' = \frac{100000}{200000} = 0.5
$$

---

**Data ke-6 (PO = 400000)**

$$
X' = \frac{400000 - 200000}{400000 - 200000}
$$

$$
X' = \frac{200000}{200000} = 1
$$

---

#### Hasil Normalisasi PO

| No | PO | PO Normalisasi |
|---|---|---|
|1|200000|0|
|2|300000|0.5|
|3|200000|0|
|4|200000|0|
|5|300000|0.5|
|6|400000|1|
|7|300000|0.5|

```python
import pandas as pd

# Dataset
data = {
    "IPK": [2,3,4,2,3,4,2],
    "PO": [200000,300000,200000,200000,300000,400000,300000]
}

df = pd.DataFrame(data)

# Rumus Min-Max
df["IPK_MinMax"] = (df["IPK"] - df["IPK"].min()) / (df["IPK"].max() - df["IPK"].min())
df["PO_MinMax"] = (df["PO"] - df["PO"].min()) / (df["PO"].max() - df["PO"].min())

print("Hasil Normalisasi Min-Max")
print(df)
```

## Normalisasi Decimal Scaling

Normalisasi Decimal Scaling adalah metode normalisasi yang dilakukan dengan cara **menggeser titik desimal pada nilai data** berdasarkan jumlah digit nilai terbesar pada atribut tersebut.

Metode ini bertujuan agar seluruh nilai data berada pada rentang **lebih kecil dari 1**.

---

#### Rumus

$$
X' = \frac{X}{10^j}
$$

**Keterangan**

- $X$ = nilai asli
- $X'$ = nilai hasil normalisasi
- $j$ = jumlah digit dari nilai maksimum

---

#### Dataset

| No | IPK | PO |
|---|---|---|
|1|2|200000|
|2|3|300000|
|3|4|200000|
|4|2|200000|
|5|3|300000|
|6|4|400000|
|7|2|300000|

---

#### Menentukan Nilai j

**Atribut IPK**

Nilai maksimum = 4  

Jumlah digit = 1  

Sehingga:

$$
j = 1
$$

---

**Atribut PO**

Nilai maksimum = 400000  

Jumlah digit = 6  

Sehingga:

$$
j = 6
$$

---

#### Perhitungan Decimal Scaling IPK

**Data ke-1**

$$
X' = \frac{2}{10^1}
$$

$$
X' = \frac{2}{10} = 0.2
$$

---

**Data ke-2**

$$
X' = \frac{3}{10}
$$

$$
X' = 0.3
$$

---

**Data ke-3**

$$
X' = \frac{4}{10}
$$

$$
X' = 0.4
$$

---

#### Perhitungan Decimal Scaling PO

**Data ke-1**

$$
X' = \frac{200000}{10^6}
$$

$$
X' = \frac{200000}{1000000} = 0.2
$$

---

**Data ke-2**

$$
X' = \frac{300000}{1000000}
$$

$$
X' = 0.3
$$

---

**Data ke-6**

$$
X' = \frac{400000}{1000000}
$$

$$
X' = 0.4
$$

---

#### Hasil Normalisasi Decimal Scaling

| No | IPK | PO | IPK DS | PO DS |
|---|---|---|---|---|
|1|2|200000|0.2|0.2|
|2|3|300000|0.3|0.3|
|3|4|200000|0.4|0.2|
|4|2|200000|0.2|0.2|
|5|3|300000|0.3|0.3|
|6|4|400000|0.4|0.4|
|7|2|300000|0.2|0.3|

```python
import pandas as pd
import numpy as np

# Dataset
data = {
    "IPK": [2,3,4,2,3,4,2],
    "PO": [200000,300000,200000,200000,300000,400000,300000]
}

df = pd.DataFrame(data)

# Menentukan j berdasarkan jumlah digit maksimum
j_ipk = len(str(df["IPK"].max()))
j_po = len(str(df["PO"].max()))

# Decimal Scaling
df["IPK_DS"] = df["IPK"] / (10**j_ipk)
df["PO_DS"] = df["PO"] / (10**j_po)

print("Hasil Normalisasi Decimal Scaling")
print(df)
```

## Normalisasi Z-Score

Normalisasi Z-Score adalah metode normalisasi yang menggunakan **rata-rata (mean)** dan **standar deviasi** untuk mengubah nilai data.

Metode ini menunjukkan seberapa jauh suatu nilai dari rata-rata dataset.

Jika nilai:

- negatif → berada di bawah rata-rata
- positif → berada di atas rata-rata
- 0 → sama dengan rata-rata

---

#### Rumus

$$
Z = \frac{X - \mu}{\sigma}
$$

**Keterangan**

- $X$ = nilai asli
- $\mu$ = rata-rata (mean)
- $\sigma$ = standar deviasi

---

#### Dataset

| No | IPK | PO |
|---|---|---|
|1|2|200000|
|2|3|300000|
|3|4|200000|
|4|2|200000|
|5|3|300000|
|6|4|400000|
|7|2|300000|

---

#### Menghitung Mean

**Mean IPK**

$$
\mu = \frac{2+3+4+2+3+4+2}{7}
$$

$$
\mu = 2.857
$$

---

**Mean PO**

$$
\mu = \frac{200000+300000+200000+200000+300000+400000+300000}{7}
$$

$$
\mu = 271428.57
$$

---

#### Menghitung Standar Deviasi

Standar deviasi menunjukkan **seberapa jauh penyebaran data dari rata-rata**.

Untuk dataset ini diperoleh:

- Standar deviasi IPK ≈ **0.83**
- Standar deviasi PO ≈ **69985**

---

#### Contoh Perhitungan

**IPK data ke-1**

$$
Z = \frac{2 - 2.857}{0.83}
$$

$$
Z ≈ -1.03
$$

---

**PO data ke-1**

$$
Z = \frac{200000 - 271428.57}{69985}
$$

$$
Z ≈ -1.02
$$

---

#### Hasil Normalisasi Z-Score

| No | IPK | PO | Z-Score IPK | Z-Score PO |
|---|---|---|---|---|
|1|2|200000|-1.03|-1.02|
|2|3|300000|0.17|0.41|
|3|4|200000|1.38|-1.02|
|4|2|200000|-1.03|-1.02|
|5|3|300000|0.17|0.41|
|6|4|400000|1.38|1.84|
|7|2|300000|-1.03|0.41|

---

#### Interpretasi

- Nilai **negatif** menunjukkan data berada **di bawah rata-rata**
- Nilai **positif** menunjukkan data berada **di atas rata-rata**
- Nilai yang semakin jauh dari 0 menunjukkan **jarak yang semakin besar dari rata-rata**

```python
import pandas as pd

# Dataset
data = {
    "IPK": [2,3,4,2,3,4,2],
    "PO": [200000,300000,200000,200000,300000,400000,300000]
}

df = pd.DataFrame(data)

# Menghitung mean dan standar deviasi
mean_ipk = df["IPK"].mean()
std_ipk = df["IPK"].std()

mean_po = df["PO"].mean()
std_po = df["PO"].std()

# Z-Score
df["IPK_ZScore"] = (df["IPK"] - mean_ipk) / std_ipk
df["PO_ZScore"] = (df["PO"] - mean_po) / std_po

print("Hasil Normalisasi Z-Score")
print(df)
```
