# Mengukur Jarak Campuran
#### Online Foods Dataset

## Data Understanding
Dataset Online Foods berisi informasi mengenai profil demografi dan perilaku pelanggan dalam layanan pemesanan makanan online. Dataset ini digunakan untuk menganalisis pola konsumsi berdasarkan faktor sosial-ekonomi. Dataset ini terdiri dari 388 data dan memiliki 13 atribut.

Tujuan utama dari penggunaan dataset ini adalah untuk mempraktikkan penghitungan kemiripan (similarity) antar pelanggan yang memiliki karakteristik data campuran (numerik dan kategorikal).

**Dataset ini terdiri dari:**
* 388 data
* 13 atribut

**4 atribut bersifat numerik, yaitu:**
* Age: umur
* Family size: jumlah anggota keluarga
* latitude: Angka desimal (koordinat lintang).
* longitude: Angka desimal (koordinat bujur).


**Atribut bersifat Kategorikal (Kualitatif)**
**Nominal (Tanpa Urutan)**
Kategori yang setara dan tidak memiliki tingkatan.

* Gender: Male, Female.
* Marital Status: Single, Married, Prefer not to say.
* Occupation: Student, Employee, Self Employeed, House wife.
* Pin code: Meskipun berupa angka, ini adalah Nominal karena fungsinya sebagai label lokasi (tidak bisa merata-ratakan kode pos).
* Output: Yes, No (Keputusan memesan).
* Unnamed: 12: Yes, No (Kolom tambahan).

**Ordinal (Memiliki Urutan/Tingkatan)**
Kategori yang memiliki jenjang atau level tertentu.

* Monthly Income: Memiliki urutan tingkat ekonomi 
* Educational Qualifications: Memiliki jenjang akademik
* Feedback: Memiliki nilai rasa atau kepuasan (Positive vs Negative).

#### Transformasi Data

Sebelum menghitung jarak, data perlu diseragamkan melalui dua tahap:

---

#### 1. Encoding Data Kategorikal

Mengubah data teks menjadi numerik.

Contoh:

- Gender: Female = 0, Male = 1  
- Occupation:  
  - Student = 0  
  - Employee = 1  
  - Self Employed = 2  
  - Housewife = 3  

---

#### 2. Normalisasi Min-Max

Dilakukan agar semua atribut numerik berada pada skala yang sama (0–1).

**Rumus:**

$$
x' = \frac{x - Min}{Max - Min}
$$

---

#### Contoh Perhitungan (Atribut Age)

Misalkan:

- Nilai minimum Age = 18  
- Nilai maksimum Age = 33  
- Usia pelanggan = 24  

$$
x' = \frac{24 - 18}{33 - 18}
$$

$$
= \frac{6}{15}
$$

$$
= 0.4
$$

---

#### Catatan

- Perhitungan normalisasi Age sudah benar secara matematis ($6/15 = 0.4$).
- Atribut **Unnamed: 12** dihapus karena merupakan duplikasi dari Output dan dapat menyebabkan redundansi dalam perhitungan jarak.

## Mengukur Jarak Tipe Data Campuran (Gower Distance)

Dataset memiliki atribut campuran (numerik dan kategorikal), sehingga digunakan **Gower Distance** untuk menghitung jarak antar data.

---

#### 1. Data yang Digunakan (3 Sampel Pertama)

Atribut yang digunakan ditunjukkan pada tabel berikut:

| Tipe Data | Atribut | Notasi |
|-----------|----------|--------|
| Numerik | Age | $p_1$ |
| Kategorikal | Gender | $p_2$ |
| Kategorikal | Occupation | $p_3$ |

---

#### Tabel Data Sampel

| Data | Age ($p_1$) | Gender ($p_2$) | Occupation ($p_3$) |
|------|------------|---------------|-------------------|
| $x_1$ | 20 | Male | Student |
| $x_2$ | 24 | Female | Student |
| $x_3$ | 22 | Male | Employee |

---

#### Representasi dalam Bentuk Vektor

$$
x_1 = (20,\; \text{Male},\; \text{Student})
$$

$$
x_2 = (24,\; \text{Female},\; \text{Student})
$$

$$
x_3 = (22,\; \text{Male},\; \text{Employee})
$$

#### 2. Rumus Dasar Gower Distance

**Rumus Umum**

$$
d(i,j) = \frac{\delta_{ij}^{(Age)} + \delta_{ij}^{(Gender)} + \delta_{ij}^{(Occupation)}}{3}
$$

Penjelasan:  
Jarak total diperoleh dari rata-rata selisih tiap atribut.

---

**Ketentuan Selisih ($\delta$)**

**A. Data Kategorikal**

Jika sama:

$$
\delta = 0
$$

Jika berbeda:

$$
\delta = 1
$$

---

**B. Data Numerik**

$$
\delta = \frac{|x_i - x_j|}{R}
$$

Dengan:

$$
R = \text{nilai maksimum} - \text{nilai minimum}
$$

Untuk Age:

$$
R = 33 - 18 = 15
$$

Artinya selisih umur dinormalisasi agar berada pada rentang 0 sampai 1.

---

#### 3. Perhitungan Manual

---

**A. Jarak antara $x_1$ dan $x_2$**

Age:

$$
\frac{|20 - 24|}{15} = \frac{4}{15} = 0.267
$$

Gender: Male vs Female → berbeda = 1  

Occupation: Student vs Student → sama = 0  

Hasil akhir:

$$
d(x_1,x_2) = \frac{0.267 + 1 + 0}{3}
$$

$$
= \frac{1.267}{3}
$$

$$
= 0.422
$$

---

**B. Jarak antara $x_1$ dan $x_3$**

Age:

$$
\frac{|20 - 22|}{15} = \frac{2}{15} = 0.133
$$

Gender: Male vs Male → sama = 0  

Occupation: Student vs Employee → berbeda = 1  

Hasil akhir:

$$
d(x_1,x_3) = \frac{0.133 + 0 + 1}{3}
$$

$$
= \frac{1.133}{3}
$$

$$
= 0.378
$$

---

**C. Jarak antara $x_2$ dan $x_3$**

Age:

$$
\frac{|24 - 22|}{15} = \frac{2}{15} = 0.133
$$

Gender: Female vs Male → berbeda = 1  

Occupation: Student vs Employee → berbeda = 1  

Hasil akhir:

$$
d(x_2,x_3) = \frac{0.133 + 1 + 1}{3}
$$

$$
= \frac{2.133}{3}
$$

$$
= 0.711
$$

---

#### 4. Matriks Jarak Gower (3x3)

$$
D =
\begin{bmatrix}
0 & 0.422 & 0.378 \\
0.422 & 0 & 0.711 \\
0.378 & 0.711 & 0
\end{bmatrix}
$$

Penjelasan:  
- Matriks simetris  
- Diagonal bernilai 0  

---

#### 5. Interpretasi Hasil

**Makna Nilai Jarak**

- Nilai mendekati 0 → sangat mirip  
- Nilai mendekati 1 → sangat berbeda  

---

**Data Paling Mirip**

$$
x_1 \text{ dan } x_3
$$

Karena memiliki jarak terkecil:

$$
0.378
$$

Mereka memiliki gender sama dan umur tidak jauh berbeda.

---

**Data Paling Berbeda**

$$
x_2 \text{ dan } x_3
$$

Karena memiliki jarak terbesar:

$$
0.711
$$

Mereka berbeda pada gender, occupation, dan usia.

---

#### Kesimpulan

- Gower Distance cocok untuk data campuran.
- Atribut numerik dinormalisasi.
- Atribut kategorikal dinilai 0 atau 1.
- Pasangan paling mirip: $x_1$ dan $x_3$.
