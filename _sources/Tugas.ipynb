# Tugas

**Missing Values Imputation dan Normalisasi Data**

Proyek ini membahas proses *data preprocessing* yang meliputi **penanganan missing values** dan **normalisasi data**.

Penanganan missing values dilakukan menggunakan metode **Weighted K-Nearest Neighbor (WKNN)**, dengan perhitungan secara **manual** dan juga menggunakan **kode program**.

Normalisasi data bertujuan untuk menyamakan skala nilai numerik. Metode yang digunakan meliputi:

- **Min-Max Normalization**
- **Z-Score Normalization**
- **Decimal Scaling**

Setiap metode disertai contoh hasil normalisasi. Implementasi dilakukan menggunakan **library `sklearn`** atau fungsi yang dibuat sendiri.
