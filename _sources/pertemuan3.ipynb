```python
# Statistik deskriptif

import pandas as pd
from scipy import stats

df = pd.read_csv("IRIS.csv")

# Ambil hanya kolom numerik (tanpa species)
df_numeric = df.select_dtypes(include=['int64','float64'])

for col in df_numeric.columns:
    print("===== Statistik untuk", col, "=====")
    print("Jumlah Data      :", df[col].count())
    print("Rata-rata        :", df[col].mean())
    print("Nilai Minimum    :", df[col].min())
    print("Q1               :", df[col].quantile(0.25))
    print("Median (Q2)      :", df[col].quantile(0.5))
    print("Q3               :", df[col].quantile(0.75))
    print("Nilai Maximum    :", df[col].max())
    print("Kemencengan      :", round(df[col].skew(), 6))

    mode = stats.mode(df[col], keepdims=True)
    print("Modus            :", mode.mode[0], "dengan jumlah", mode.count[0])

    print("Standar Deviasi  :", round(df[col].std(), 6))
    print("Variansi         :", round(df[col].var(), 6))
    print("\n")
```

### Feature Statistic
Feature Statistics adalah fitur di Orange yang digunakan untuk melihat gambaran umum dari setiap atribut dalam dataset, seperti rata-rata, nilai tengah, nilai minimum dan maksimum, penyebaran data, serta grafik distribusinya. Dengan fitur ini, kita bisa memahami bagaimana pola dan sebaran data, melihat perbedaan antar kelas, serta memastikan tidak ada data yang kosong, sehingga membantu dalam menentukan fitur mana yang paling penting sebelum melakukan proses analisis atau klasifikasi.

Pada tahap ini kita melihat ringkasan statistik dari setiap kolom data.
Di sini ditampilkan informasi seperti:
* Mean : nilai rata-rata
* Mode : nilai yang paling sering muncul
* Median : nilai tengah
* Min & Max : nilai terkecil dan terbesar
* Dispersion : tingkat penyebaran data
* Missing : jumlah data yang kosong (di sini tidak ada)

Selain itu, terlihat juga grafik kecil (histogram) yang menunjukkan sebaran data untuk tiap jenis bunga (warna biru, merah, hijau).

### Scatter Plot
Scatter plot adalah grafik yang digunakan untuk melihat hubungan antara dua fitur dalam dataset dengan menampilkan data dalam bentuk titik-titik. Pada gambar tersebut, sumbu X menunjukkan sepal length dan sumbu Y menunjukkan sepal width, sedangkan setiap titik mewakili satu data bunga dan warnanya menunjukkan kelasnya (Iris-setosa, Iris-versicolor, dan Iris-virginica). Dari grafik terlihat bahwa Iris-setosa cukup terpisah dari kelas lainnya karena memiliki ukuran yang berbeda, sedangkan Iris-versicolor dan Iris-virginica masih saling tumpang tindih. Hal ini menunjukkan bahwa dua fitur ini sudah bisa membedakan sebagian kelas, tetapi belum sepenuhnya memisahkan semua kelas dengan jelas.

### Box Plot
Box plot adalah grafik yang digunakan untuk melihat penyebaran data dan membandingkan beberapa kelompok dengan mudah. Pada gambar tersebut, setiap kotak mewakili data panjang sepal dari masing-masing jenis bunga Iris. Garis di tengah kotak menunjukkan nilai tengah (median), sedangkan bagian kotak menunjukkan sebagian besar data berada di rentang tersebut. Garis yang memanjang ke atas dan ke bawah (whiskers) menunjukkan nilai terendah dan tertinggi. Dari posisi kotaknya terlihat bahwa Iris-virginica memiliki sepal paling panjang dibandingkan jenis lainnya. Nilai uji ANOVA yang sangat kecil (p = 0.000) menunjukkan bahwa perbedaan ukuran antar ketiga jenis bunga tersebut benar-benar signifikan dan bukan terjadi secara kebetulan.

### Correlations
Correlation (korelasi) adalah angka yang menunjukkan seberapa kuat hubungan antara dua variabel, dengan nilai antara -1 sampai 1. Jika nilainya mendekati +1, berarti hubungannya positif dan searah (jika satu naik, yang lain ikut naik). Jika mendekati -1, berarti hubungannya berlawanan arah. Jika mendekati 0, berarti hampir tidak ada hubungan. Pada data Iris, hubungan antara petal length dan petal width sangat tinggi (+0,963), artinya semakin panjang kelopak bunga, biasanya semakin lebar juga—keduanya tumbuh hampir sebanding. Sebaliknya, korelasi antara sepal length dan sepal width sangat rendah (-0,109), yang berarti panjang dan lebar sepal tidak terlalu berhubungan dan cenderung berdiri sendiri.

### Rank
Rank di Orange adalah fitur yang digunakan untuk menilai dan mengurutkan atribut berdasarkan seberapa penting atau seberapa besar pengaruhnya terhadap target (kelas). Rank menghitung skor setiap fitur menggunakan metode tertentu (seperti Information Gain, Chi-Square, atau Gain Ratio), lalu menampilkannya dalam bentuk peringkat dari yang paling berpengaruh sampai yang paling rendah. Dengan Rank, kita bisa mengetahui fitur mana yang paling relevan untuk membedakan kelas dan memilih atribut terbaik sebelum membuat model klasifikasi.

### Pivot Table
Pivot Table di Orange adalah fitur yang digunakan untuk merangkum dan menyusun ulang data agar lebih mudah dianalisis. Dengan Pivot Table, kita bisa mengelompokkan data berdasarkan kategori tertentu, lalu melihat jumlah (count), rata-rata (mean), atau perhitungan lainnya dalam bentuk tabel ringkas. Fitur ini membantu kita melihat pola, perbandingan antar kelompok, serta memahami distribusi data tanpa harus melihat satu per satu isi dataset secara detail.
