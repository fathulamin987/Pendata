# Proyek UTS

**Missing Values Imputation dan Normalisasi Data**

Pada proyek UTS ini dilakukan pembahasan mengenai **penanganan missing values** dan **normalisasi data** sebagai bagian dari proses *data preprocessing*.

Penanganan missing values dilakukan menggunakan metode **Weighted K-Nearest Neighbor Imputation (WKNN)**. Perhitungan metode WKNN dilakukan secara **manual untuk memahami proses perhitungannya**, serta menggunakan **kode program untuk menghitung nilai imputasi**.

Selain itu, proyek ini juga membahas **normalisasi data**, yang bertujuan untuk menyamakan skala nilai pada data numerik. Beberapa metode normalisasi yang dijelaskan antara lain:

- **Min-Max Normalization**
- **Z-Score Normalization**
- **Decimal Scaling**

Setiap metode akan diberikan **contoh hasil normalisasi data**.

Implementasi normalisasi dilakukan menggunakan **library `sklearn`** atau dengan **membuat fungsi normalisasi sendiri**.
