## Data understanding
Data Understanding adalah salah satu tahap dalam proses data mining yang bertujuan untuk memahami data sebelum dianalisis lebih lanjut. Pada tahap ini, data diperiksa untuk mengetahui isi, struktur, dan kondisinya agar siap digunakan dalam proses berikutnya.

Pada Dataset Iris, Data Understanding dilakukan dengan melihat atribut petal, yaitu petal_length dan petal_width. Dari atribut tersebut dapat diamati pola, sebaran data, dan perbedaan antar spesies. Dengan memahami bagian petal ini, kita dapat mengetahui bagaimana ciri-ciri bunga Iris berbeda satu sama lain sebelum dilakukan proses analisis atau pemodelan.

## Dataset Iris Flower
Dataset Iris Flower merupakan dataset multivariat yang diperkenalkan oleh ahli statistika dan biologi Inggris, Ronald Fisher, pada tahun 1936. Dataset bunga Iris ini sangat terkenal di dunia Machine Learning dan sering digunakan untuk proses klasifikasi.

Dataset Iris Flower merupakan data terstruktur karena dinyatakan dalam bentuk tabel yang terdiri dari atribut (kolom) dan baris.

### Dataset ini terdiri dari:
* 150 data
* 5 atribut

### Empat atribut bersifat numerik, yaitu:
* sepal_length
* sepal_width
* petal_length
* petal_width

### Satu atribut bersifat kategorikal, yaitu:
* species

### Pada atribut species terdapat tiga kategori, yaitu:
* iris-setosa
* iris-versicolor
* iris-virginica

```python
# Statistik deskriptif

import pandas as pd
from scipy import stats

df = pd.read_csv("IRIS.csv")

# Ambil hanya kolom numerik (tanpa species)
df_numeric = df.select_dtypes(include=['int64','float64'])

for col in df_numeric.columns:
    print("===== Statistik untuk", col, "=====")
    print("Jumlah Data      :", df[col].count())
    print("Rata-rata        :", df[col].mean())
    print("Nilai Minimum    :", df[col].min())
    print("Q1               :", df[col].quantile(0.25))
    print("Median (Q2)      :", df[col].quantile(0.5))
    print("Q3               :", df[col].quantile(0.75))
    print("Nilai Maximum    :", df[col].max())
    print("Kemencengan      :", round(df[col].skew(), 6))

    mode = stats.mode(df[col], keepdims=True)
    print("Modus            :", mode.mode[0], "dengan jumlah", mode.count[0])

    print("Standar Deviasi  :", round(df[col].std(), 6))
    print("Variansi         :", round(df[col].var(), 6))
    print("\n")
```
